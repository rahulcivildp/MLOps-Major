```python
plt.figure(figsize=(8, 5))

sns.histplot(df['target'], bins=40, discrete=True)

plt.title("Olivetti Faces Class Distribution")
plt.xlabel("Subject ID")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6, 6))

plt.imshow(X[0].reshape(64, 64), cmap="gray")
plt.title(f"Example Olivetti Face (Subject {y[0]})")
plt.axis("off")

plt.show()
```

```python
# Use a smaller feature subset for correlation heatmap to keep plotting fast.
subset = df.iloc[:, :25].copy()
subset['target'] = df['target']

corr = subset.corr()

plt.figure(figsize=(12, 10))
sns.heatmap(
    corr,
    annot=False,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap (first 25 pixels + target)")

plt.show()
```
